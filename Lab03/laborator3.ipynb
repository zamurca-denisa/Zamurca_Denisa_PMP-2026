## a) Independențele din rețea

Structura rețelei este:

$$
S \rightarrow O
$$

$$
S \rightarrow L
$$

$$
S \rightarrow M
$$

$$
L \rightarrow M
$$

Variabilele $O$ și $L$ sunt independente condiționat de $S$:

$$
O \perp L \mid S
$$

De asemenea, $O$ și $M$ sunt independente condiționat de $S$:

$$
O \perp M \mid S
$$

În particular, dacă știm valorile lui $S$ și $L$, atunci $O$ nu mai aduce informație suplimentară despre $M$:

$$
M \perp O \mid S,L
$$

## b) Clasificarea e-mail-urilor

Pentru clasificare comparăm cele două cazuri:

$$
P(S=1,O,L,M)
$$

și

$$
P(S=0,O,L,M)
$$

Rețeaua are factorizarea:

$$
P(S,O,L,M)
=
P(S)\cdot P(O\mid S)\cdot P(L\mid S)\cdot P(M\mid S,L)
$$

Pentru fiecare combinație a atributelor $O$, $L$ și $M$, calculăm cele două valori și alegem clasa cu probabilitatea mai mare.

### Exemplu: $O=1$, $L=1$, $M=1$

Pentru spam:

$$
P(S=1,O=1,L=1,M=1)
=
P(S=1)\cdot P(O=1\mid S=1)\cdot P(L=1\mid S=1)\cdot P(M=1\mid S=1,L=1)
$$

$$
=0.4\cdot0.7\cdot0.8\cdot0.9
$$

$$
=0.2016
$$

Pentru non-spam:

$$
P(S=0,O=1,L=1,M=1)
=
P(S=0)\cdot P(O=1\mid S=0)\cdot P(L=1\mid S=0)\cdot P(M=1\mid S=0,L=1)
$$

$$
=0.6\cdot0.1\cdot0.3\cdot0.6
$$

$$
=0.0108
$$

Deoarece:

$$
0.2016 > 0.0108
$$

e-mailul este clasificat drept:

$$
\boxed{\text{SPAM}}
$$

```python
# ==================================================
# EXERCITIUL 3 - RETEA BAYESIANA PENTRU SPAM
# ==================================================

# Importam instrumentele pentru reteaua Bayesiana
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination

# Pentru desenarea grafului
import networkx as nx
import matplotlib.pyplot as plt

# Pentru generarea tuturor combinatiilor posibile de 0 si 1
from itertools import product


# ==================================================
# 1. DEFINIM RETEAUA BAYESIANA
# ==================================================

# S = mesaj spam
# O = mesajul contine cuvantul "oferta"
# L = mesajul contine linkuri
# M = mesajul are lungime mare

# Fiecare variabila are doua valori:
# 0 = nu
# 1 = da

# Construim reteaua prin stabilirea sagetilor.
#
# ("S", "O") inseamna S -> O

model = DiscreteBayesianNetwork([
    ("S", "O"),
    ("S", "L"),
    ("S", "M"),
    ("L", "M")
])

print("Muchiile retelei:")
print(list(model.edges()))


# ==================================================
# 2. DEFINIM PROBABILITATILE PENTRU S
# ==================================================

# Din enunt:
# P(S = 0) = 0.6
# P(S = 1) = 0.4

cpd_S = TabularCPD(
    variable="S",
    variable_card=2,
    values=[
        [0.6],  # S = 0: mesajul nu este spam
        [0.4]   # S = 1: mesajul este spam
    ]
)


# ==================================================
# 3. DEFINIM PROBABILITATILE PENTRU O
# ==================================================

# O depinde de S.
#
# Din enunt:
# P(O = 1 | S = 0) = 0.1
# P(O = 1 | S = 1) = 0.7
#
# Completam si probabilitatile pentru O = 0:
#
# P(O = 0 | S = 0) = 1 - 0.1 = 0.9
# P(O = 0 | S = 1) = 1 - 0.7 = 0.3
#
# Coloane: S = 0, S = 1
# Randuri: O = 0, O = 1

cpd_O = TabularCPD(
    variable="O",
    variable_card=2,
    values=[
        [0.9, 0.3],  # O = 0
        [0.1, 0.7]   # O = 1
    ],
    evidence=["S"],
    evidence_card=[2]
)


# ==================================================
# 4. DEFINIM PROBABILITATILE PENTRU L
# ==================================================

# L depinde de S.
#
# Din enunt:
# P(L = 1 | S = 0) = 0.3
# P(L = 1 | S = 1) = 0.8
#
# Completam:
# P(L = 0 | S = 0) = 0.7
# P(L = 0 | S = 1) = 0.2
#
# Coloane: S = 0, S = 1
# Randuri: L = 0, L = 1

cpd_L = TabularCPD(
    variable="L",
    variable_card=2,
    values=[
        [0.7, 0.2],  # L = 0
        [0.3, 0.8]   # L = 1
    ],
    evidence=["S"],
    evidence_card=[2]
)


# ==================================================
# 5. DEFINIM PROBABILITATILE PENTRU M
# ==================================================

# M depinde de doua variabile: S si L.
#
# Pentru evidence=["S", "L"], coloanele sunt:
#
# Coloana 1: S = 0, L = 0
# Coloana 2: S = 0, L = 1
# Coloana 3: S = 1, L = 0
# Coloana 4: S = 1, L = 1
#
# Din enunt:
#
# P(M = 1 | S = 0, L = 0) = 0.2
# P(M = 1 | S = 0, L = 1) = 0.6
# P(M = 1 | S = 1, L = 0) = 0.5
# P(M = 1 | S = 1, L = 1) = 0.9
#
# Probabilitatile pentru M = 0 se calculeaza cu 1 - P(M = 1).

cpd_M = TabularCPD(
    variable="M",
    variable_card=2,
    values=[
        [0.8, 0.4, 0.5, 0.1],  # M = 0
        [0.2, 0.6, 0.5, 0.9]   # M = 1
    ],
    evidence=["S", "L"],
    evidence_card=[2, 2]
)


# ==================================================
# 6. ADAUGAM TABELELE IN MODEL
# ==================================================

model.add_cpds(
    cpd_S,
    cpd_O,
    cpd_L,
    cpd_M
)

# Verificam daca modelul este construit corect
print("\nModelul este corect:", model.check_model())


# ==================================================
# 7. AFISAM TABELELE DE PROBABILITATE
# ==================================================

print("\nTabelul pentru S:")
print(cpd_S)

print("\nTabelul pentru O:")
print(cpd_O)

print("\nTabelul pentru L:")
print(cpd_L)

print("\nTabelul pentru M:")
print(cpd_M)


# ==================================================
# 8. DESENAM GRAFUL RETELEI
# ==================================================

# Stabilim pozitiile nodurilor pentru un graf clar
pozitii = {
    "S": (0, 1),
    "O": (1, 2),
    "L": (1, 0),
    "M": (2, 1)
}

plt.figure(figsize=(7, 5))

nx.draw(
    model,
    pos=pozitii,
    with_labels=True,
    node_color="lightblue",
    node_size=2500,
    font_size=14,
    font_weight="bold",
    arrows=True
)

plt.title("Reteaua Bayesiana pentru clasificarea spam")

# Salvam graful ca imagine
plt.savefig(
    "retea_bayesiana_spam.png",
    dpi=150,
    bbox_inches="tight"
)

plt.show()


# ==================================================
# 9. a) INDEPENDENTELE DIN RETEA
# ==================================================

# Exemple de independente care rezulta din graf:
#
# O si L sunt independente conditionat de S:
# O ⟂ L | S
#
# O si M sunt independente conditionat de S:
# O ⟂ M | S
#
# Interpretare:
# Daca stim deja daca mesajul este spam sau nu,
# observarea lui O nu mai ofera informatie suplimentara
# despre L sau M, conform acestui model.

print("\nExemple de independente conditionate:")
print("O este independent de L, cunoscand S.")
print("O este independent de M, cunoscand S.")

# Putem afisa si independentele gasite de biblioteca
print("\nIndependentele afisate de pgmpy:")
print(model.get_independencies())


# ==================================================
# 10. b) CLASIFICAREA MESAJELOR
# ==================================================

# Cream obiectul care face inferenta in reteaua Bayesiana
inferenta = VariableElimination(model)

print("\nClasificarea pentru toate combinatiile posibile:")
print("O = oferta, L = linkuri, M = mesaj lung")
print()

# Generam toate combinatiile posibile de valori pentru O, L si M.
#
# Sunt 2 * 2 * 2 = 8 combinatii.
for O, L, M in product([0, 1], repeat=3):

    # Calculam distributia lui S, stiind valorile lui O, L si M
    rezultat = inferenta.query(
        variables=["S"],
        evidence={
            "O": O,
            "L": L,
            "M": M
        },
        show_progress=False
    )

    # S = 0 inseamna non-spam
    probabilitate_non_spam = rezultat.values[0]

    # S = 1 inseamna spam
    probabilitate_spam = rezultat.values[1]

    # Alegem clasa cu probabilitatea mai mare.
    # Pragul 0.5 corespunde costurilor egale pentru greseli.
    if probabilitate_spam > 0.5:
        clasificare = "SPAM"
    else:
        clasificare = "NON-SPAM"

    print(
        f"O={O}, L={L}, M={M} | "
        f"P(spam)={probabilitate_spam:.4f} | "
        f"P(non-spam)={probabilitate_non_spam:.4f} | "
        f"Clasificare: {clasificare}"
    )


# ==================================================
# 11. EXEMPLU INDIVIDUAL
# ==================================================

# Exemplu:
# O = 1 -> contine cuvantul "oferta"
# L = 1 -> contine linkuri
# M = 1 -> mesajul este lung

rezultat_exemplu = inferenta.query(
    variables=["S"],
    evidence={
        "O": 1,
        "L": 1,
        "M": 1
    },
    show_progress=False
)

print("\nExemplu: O=1, L=1, M=1")
print(rezultat_exemplu)
```
