```python
import numpy as np
from hmmlearn.hmm import CategoricalHMM

stari = ["1", "2", "3", "4"]
culori = ["red", "green", "blue"]

pi = np.array([1/4, 1/4, 1/4, 1/4])

A = np.array([
    [0.20, 0.40, 0.0, 0.40],
    [0.40, 0.20, 0.40, 0.0],
    [0.0, 0.40, 0.20, 0.40],
    [0.40, 0.0, 0.40, 0.20]
])


B = np.array([
    [0.70, 0.20, 0.10],
    [0.20, 0.60, 0.20],
    [0.10, 0.30, 0.60],
    [0.50, 0.20, 0.30]
])

model = CategoricalHMM(
    n_components=4,
    n_features=3,
    init_params=""
)

model.startprob_ = pi
model.transmat_ = A
model.emissionprob_ = B

print("Stări:", stari)
print("Observații:", culori)
```

```python
culori_observate = [
    "red", "green", "blue", "blue", "red"
]

cod_culoare = {
    "red": 0,
    "green": 1,
    "blue": 2
}

observatii = np.array(
    [cod_culoare[culoare] for culoare in culori_observate]
).reshape(-1, 1)


log_prob_comuna, secventa_stari = model.decode(
    observatii,
    algorithm="viterbi"
)

log_prob_note = model.score(observatii)

prob_comuna = np.exp(log_prob_comuna)


print("Rezultatul pentru fiecare test:\n")

for i in range(len(culori_observate)):
    print(
        f"Testul {i + 1}: "
        f"nota = {culori_observate[i]}, "
        f"dificultate estimată = {stari[secventa_stari[i]]}"
    )

print("\nSecvența:")
print(" → ".join(stari[i] for i in secventa_stari))
```

Subiectul 2. Retele Markov

subpunctul a)

```python
import math
import itertools

variabile = ["A", "B", "C", "D", "E"]

clici_maximale = [
    ["A", "B"],
    ["B", "C"],
    ["C", "D"],
    ["D", "E"]
]

beta = 1.0

def calculeaza_potential(clica, configuratie, beta):

    suma_perechi = 0

    for var1, var2 in itertools.combinations(clica, 2):

        suma_perechi += (
            configuratie[var1] * configuratie[var2]
        )

    potential = math.exp(beta * suma_perechi)

    return potential

configuratii = list(
    itertools.product([0, 1], repeat=5)
)

ponderi = []

for valori in configuratii:
    configuratie = dict(zip(variabile, valori))
    pondere = 1.0
    for clica in clici_maximale:

        potential = calculeaza_potential(
            clica,
            configuratie,
            beta
        )

        pondere *= potential
    ponderi.append(pondere)
Z = sum(ponderi)

print("Valoarea lui beta:", beta)
print("Constanta Z:", Z)

probabilitati = []

for pondere in ponderi:
    probabilitate = pondere / Z

    probabilitati.append(probabilitate)

print("\nProbabilitatile tuturor configuratiilor:\n")

for configuratie, probabilitate in zip(
    configuratii,
    probabilitati
):
    print(
        configuratie,
        "->",
        round(probabilitate, 6)
    )
```
