```python
import csv
import random

# Lista in care vom salva studentii cititi din fisier
studenti = []

# Deschidem fisierul CSV
with open("studenti.csv", "r") as fisier:
    cititor = csv.reader(fisier)

    # Citim fiecare student si il adaugam in lista
    for rand in cititor:
        studenti.append(rand[0])

# Numarul de studenti pe care vrem sa ii alegem
numar_studenti = 3

# Alegem aleatoriu studentii, fara repetare
studenti_alesi = random.sample(studenti, numar_studenti)

# Afisam studentii alesi
print("Studentii alesi sunt:")

for student in studenti_alesi:
    print(student)
```
