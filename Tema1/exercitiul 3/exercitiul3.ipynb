```python
import numpy as np
import matplotlib.pyplot as plt

# Numarul de valori pe care vrem sa le generam
numar_valori = 10000

# Lista in care vom salva timpii de servire
valori_X = []

# Probabilitatile ca un client sa fie preluat de fiecare frizer
p1 = 3 / 13
p2 = 6 / 13
p3 = 4 / 13

# Generam 10000 de valori pentru X
for i in range(numar_valori):

    # Alegem un numar aleatoriu intre 0 si 1
    numar = np.random.random()

    # Daca numarul este mai mic decat p1,
    # clientul este preluat de primul frizer
    if numar < p1:

        X = np.random.exponential(scale=1 / 3)

    # Daca nu a fost ales primul frizer,
    # verificam daca este ales al doilea
    elif numar < p1 + p2:

        X = np.random.exponential(scale=1 / 6)

    # Altfel, clientul este preluat de al treilea frizer
    else:

        X = np.random.exponential(scale=1 / 4)

    # Adaugam valoarea obtinuta in lista
    valori_X.append(X)

# Transformam lista intr-un array NumPy
valori_X = np.array(valori_X)

# Calculam media
media = np.mean(valori_X)

# Calculam deviatia standard
deviatia_standard = np.std(valori_X)

# Afisam rezultatele
print("Media aproximativa a lui X este:", media)
print("Deviatia standard aproximativa a lui X este:", deviatia_standard)
```

```python
# Reprezentam aproximativ densitatea distributiei lui X

plt.hist(valori_X, bins=50, density=True)

plt.xlabel("Timp de servire X")
plt.ylabel("Densitate")
plt.title("Distributia aproximativa a timpului de servire")

plt.show()
```
