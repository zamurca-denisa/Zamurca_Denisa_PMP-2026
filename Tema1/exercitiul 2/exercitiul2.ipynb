```python
import random

# Initializam numarul de pasi si suma totala
N = 0
S = 0

# Variabila care ne spune daca jocul s-a terminat
joc_terminat = False

# Repetam aruncarea monedei pana cand apare stema
while joc_terminat == False:

    # La fiecare aruncare crestem numarul de pasi
    N = N + 1

    # Alegem aleatoriu intre stema si ban
    moneda = random.choice(["stema", "ban"])

    # Daca apare stema
    if moneda == "stema":

        # Aruncam zarul
        zar = random.randint(1, 6)

        # Al doilea jucator ii da primului suma z - 3
        suma = zar - 3
        S = S + suma

        # Jocul se termina
        joc_terminat = True

    # Daca apare ban
    else:

        # Primul jucator ii da celui de-al doilea 0.5 dolari
        # De aceea scadem 0.5 din S
        S = S - 0.5

# Afisam rezultatele
print("Numarul de pasi N este:", N)
print("Suma totala S este:", S)
```

```python
import random

# Functie care simuleaza un singur joc
def joc():

    # Initializam numarul de pasi si suma
    N = 0
    S = 0

    joc_terminat = False

    # Jocul continua pana la aparitia stemei
    while joc_terminat == False:

        N = N + 1

        # Aruncam moneda
        moneda = random.choice(["stema", "ban"])

        # Daca apare stema
        if moneda == "stema":

            # Aruncam zarul
            zar = random.randint(1, 6)

            # Calculam suma platita de jucatorul 2
            suma = zar - 3
            S = S + suma

            # Oprim jocul
            joc_terminat = True

        # Daca apare ban
        else:

            # Jucatorul 1 plateste 0.5 dolari
            S = S - 0.5

    # Returnam numarul de pasi si suma finala
    return N, S
```

```python
# Lista in care salvam valorile lui S
rezultate_S = []

# Numarul de jocuri pe care vrem sa le simulam
numar_jocuri = 100000

# Simulam jocul de 100000 de ori
for i in range(numar_jocuri):

    N, S = joc()

    # Salvam rezultatul lui S
    rezultate_S.append(S)
```

```python
# Calculam suma tuturor rezultatelor
suma_totala = 0

for rezultat in rezultate_S:
    suma_totala = suma_totala + rezultat

# Calculam media
media = suma_totala / len(rezultate_S)

# Afisam media
print("Media aproximativa a lui S este:", media)
```

```python
# În urma simulărilor, observăm că media lui S este aproximativ 0.
```

```python
import matplotlib.pyplot as plt

# Construim histograma valorilor lui S
plt.hist(rezultate_S, bins=30)

# Adaugam numele axelor si titlul graficului
plt.xlabel("S")
plt.ylabel("Frecventa")
plt.title("Distributia sumei S")

# Afisam graficul
plt.show()
```

```python
import random

# Functie care simuleaza jocul pentru o probabilitate p
# de aparitie a stemei
def joc_masluit(p):

    N = 0
    S = 0

    joc_terminat = False

    while joc_terminat == False:

        N = N + 1

        # Generam un numar aleatoriu intre 0 si 1
        numar = random.random()

        # Consideram ca apare stema daca numarul este mai mic decat p
        if numar < p:

            # Aruncam zarul
            zar = random.randint(1, 6)

            # Calculam suma platita de jucatorul 2
            suma = zar - 3
            S = S + suma

            # Jocul se termina
            joc_terminat = True

        else:

            # A aparut ban, deci jucatorul 1 plateste 0.5 dolari
            S = S - 0.5

    return N, S
```

```python
# Pentru p = 0.3
```

```python
# Lista pentru valorile lui S
rezultate_S_03 = []

# Numarul de simulari
numar_jocuri = 100000

# Simulam jocurile pentru p = 0.3
for i in range(numar_jocuri):

    N, S = joc_masluit(0.3)

    rezultate_S_03.append(S)


# Calculam suma tuturor rezultatelor
suma_totala = 0

for rezultat in rezultate_S_03:
    suma_totala = suma_totala + rezultat

# Calculam media
media_03 = suma_totala / len(rezultate_S_03)

print("Media aproximativa pentru p = 0.3 este:", media_03)
```

```python
import matplotlib.pyplot as plt

# Construim histograma
plt.hist(rezultate_S_03, bins=30)

# Adaugam informatii graficului
plt.xlabel("S")
plt.ylabel("Frecventa")
plt.title("Distributia lui S pentru p = 0.3")

# Afisam graficul
plt.show()
```

```python
# Pentru p = 0.3, stema apare mai rar. Din această cauză, jocul durează în general mai mult, iar primul jucător ajunge să plătească de mai multe ori suma de 0.5 dolari. Media lui S devine astfel negativă.
```
