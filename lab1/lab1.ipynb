```python
import random


# ============================================
# DATELE PROBLEMEI
# ============================================

# Vom repeta experimentul de 100.000 de ori
numar_experimente = 100_000

# Aici numaram de cate ori extragem o bila rosie
numar_rosii = 0


# ============================================
# a) si b) SIMULAREA EXPERIMENTULUI
# ============================================

for i in range(numar_experimente):

    # La inceputul fiecarui experiment refacem urna:
    # 3 bile rosii, 4 albastre si 2 negre
    urna = (
        ["rosie"] * 3
        + ["albastra"] * 4
        + ["neagra"] * 2
    )

    # Aruncam zarul: obtinem un numar intre 1 si 6
    zar = random.randint(1, 6)

    # Daca zarul este 2, 3 sau 5, adaugam o bila neagra
    if zar in [2, 3, 5]:
        urna.append("neagra")

    # Daca zarul este 6, adaugam o bila rosie
    elif zar == 6:
        urna.append("rosie")

    # In celelalte cazuri, adica 1 sau 4, adaugam o bila albastra
    else:
        urna.append("albastra")

    # Extragem aleator o bila din urna
    bila_extrasa = random.choice(urna)

    # Daca bila extrasa este rosie, crestem contorul
    if bila_extrasa == "rosie":
        numar_rosii += 1


# Calculam probabilitatea estimata prin simulare
probabilitate_estimata = numar_rosii / numar_experimente


# ============================================
# c) CALCULUL PROBABILITATII TEORETICE
# ============================================

# Cazul 1: zarul este 2, 3 sau 5
# Probabilitatea acestui caz este 3/6.
# Adaugam o bila neagra, deci avem 3 bile rosii din 10.
caz_prim = (3 / 6) * (3 / 10)

# Cazul 2: zarul este 6
# Probabilitatea acestui caz este 1/6.
# Adaugam o bila rosie, deci avem 4 bile rosii din 10.
caz_sase = (1 / 6) * (4 / 10)

# Cazul 3: zarul este 1 sau 4
# Probabilitatea acestui caz este 2/6.
# Adaugam o bila albastra, deci avem 3 bile rosii din 10.
caz_albastra = (2 / 6) * (3 / 10)

# Adunam probabilitatile celor trei cazuri
probabilitate_teoretica = caz_prim + caz_sase + caz_albastra


# ============================================
# AFISAREA SI COMPARAREA REZULTATELOR
# ============================================

# Calculam diferenta dintre estimare si rezultatul teoretic
diferenta = abs(
    probabilitate_estimata - probabilitate_teoretica
)

print("Numar de experimente:", numar_experimente)
print("Numar de bile rosii extrase:", numar_rosii)

print("\nProbabilitatea estimata:", probabilitate_estimata)
print("Probabilitatea teoretica:", probabilitate_teoretica)
print("Diferenta:", diferenta)
```
