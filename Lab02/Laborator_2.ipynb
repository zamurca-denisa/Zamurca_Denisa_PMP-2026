# a)

Notăm cu $B$ evenimentul că persoana este bolnavă și cu $+$ evenimentul că testul este pozitiv.

Avem:

$$
P(B) = 0.01
$$

$$
P(\neg B) = 1 - P(B) = 0.99
$$

Sensibilitatea testului este:

$$
P(+ \mid B) = 0.95
$$

Specificitatea este:

$$
P(- \mid \neg B) = 0.90
$$

Prin urmare, probabilitatea unui rezultat fals pozitiv este:

$$
P(+ \mid \neg B) = 1 - 0.90 = 0.10
$$

Aplicăm formula lui Bayes:

$$
P(B \mid +) =
\frac{P(+ \mid B)P(B)}
{P(+ \mid B)P(B) + P(+ \mid \neg B)P(\neg B)}
$$

Înlocuind valorile:

$$
P(B \mid +) =
\frac{0.95 \cdot 0.01}
{0.95 \cdot 0.01 + 0.10 \cdot 0.99}
$$

$$
P(B \mid +) =
\frac{0.0095}
{0.0095 + 0.099}
$$

$$
P(B \mid +) =
\frac{0.0095}{0.1085}
\approx 0.0876
$$

Deci:

$$
\boxed{P(B \mid +) \approx 8.76\%}
$$

Rezultatul este relativ mic deoarece boala apare doar la $1\%$ din populație. Deși testul are o sensibilitate mare, există o probabilitate de $10\%$ ca o persoană sănătoasă să primească un rezultat pozitiv. Cum persoanele sănătoase reprezintă $99\%$ din populație, numărul rezultatelor fals pozitive poate fi mult mai mare decât numărul rezultatelor pozitive corecte.

b)

Modelăm în Python probabilitatea ca o persoană să fie bolnavă, știind că testul a fost pozitiv, folosind formula lui Bayes.

Comparăm apoi rezultatul obținut cu valoarea teoretică de la punctul a), aproximativ $8.76\%$.

```python
p_boala = 0.01
p_nu_boala = 1 - p_boala
p_pozitiv_daca_boala = 0.95
specificitate = 0.90
p_pozitiv_daca_nu_boala = 1 - specificitate

p_boala_daca_pozitiv = (
    p_pozitiv_daca_boala * p_boala
) / (
    p_pozitiv_daca_boala * p_boala
    + p_pozitiv_daca_nu_boala * p_nu_boala
)

print("Probabilitatea ca persoana sa fie bolnava daca testul este pozitiv este:")
print(p_boala_daca_pozitiv)

print("In procente:")
print(p_boala_daca_pozitiv * 100, "%")
```

Rezultatul obținut în Python este aproximativ:
8.76\% .

Acesta coincide cu rezultatul teoretic obținut la punctul a).

Dorim ca probabilitatea ca o persoană să fie bolnavă, știind că testul este pozitiv, să fie de $50\%$:

$$
P(B \mid +) = 0.5
$$

Notăm specificitatea testului cu $s$.

Atunci probabilitatea unui rezultat fals pozitiv este:

$$
P(+ \mid \neg B) = 1 - s
$$

Folosim formula lui Bayes:

$$
P(B \mid +) =
\frac{P(+ \mid B)P(B)}
{P(+ \mid B)P(B) + P(+ \mid \neg B)P(\neg B)}
$$

Înlocuim valorile cunoscute:

$$
0.5 =
\frac{0.95 \cdot 0.01}
{0.95 \cdot 0.01 + (1-s)\cdot 0.99}
$$

Calculăm:

$$
0.5 =
\frac{0.0095}
{0.0095 + 0.99(1-s)}
$$

Înmulțim cu numitorul:

$$
0.5(0.0095 + 0.99(1-s)) = 0.0095
$$

Împărțim prin $0.5$:

$$
0.0095 + 0.99(1-s) = 0.019
$$

Rezultă:

$$
0.99(1-s) = 0.0095
$$

$$
1-s = \frac{0.0095}{0.99}
$$

$$
1-s \approx 0.009596
$$

Prin urmare:

$$
s \approx 1 - 0.009596
$$

$$
s \approx 0.990404
$$

Deci specificitatea minimă necesară este aproximativ:

$$
\boxed{s \approx 99.04\%}
$$

Testul trebuie deci să aibă o specificitate de cel puțin aproximativ $99.04\%$ pentru ca, în cazul unui rezultat pozitiv, probabilitatea ca persoana să fie într-adevăr bolnavă să ajungă la $50\%$.
